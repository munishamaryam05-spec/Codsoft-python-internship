```python
def todo_list():
    tasks = []
    while True:
        print("====================")
        print("    TO-DO LIST      ")
        print("====================")
        print("1. View Tasks")
        print("2. Add Task")
        print("3. Update Task")
        print("4. Mark Task as Done")
        print("5. Delete Task")
        print("6. Exit")
        
        choice = input("Enter your choice: ")
        
        if choice == "1":
            print("-----------------------")
            print("    YOUR TASK LIST     ")
            print("-----------------------")
            if not tasks:
                print("No tasks found!")
            else:
                print("\nYour Tasks:\n")
                for i, task in enumerate(tasks, 1):
                    status = "Done" if task["done"] else "Pending"
                    print(f"{i}. {task['name']:<25} [{status}]\n")
        
        elif choice == "2":
            print("-----------------------")
            print("     ADD NEW TASK      ")
            print("-----------------------")
            task_name = input("Enter task: ")
            tasks.append({"name": task_name, "done": False})
            print(f"Task '{task_name}' added Successfully!")
    
        elif choice == "3":
            print("-----------------------------")
            print("    UPDATE EXISTING TASK     ")
            print("-----------------------------")
            if not tasks:
                print("No tasks to update!")
            else:
                for i, task in enumerate(tasks, 1):
                    print(f"{i}. {task['name']}")
                num = int(input("Enter task number to update: ")) - 1
                new_name = input("Enter new task name: ")
                tasks[num]["name"] = new_name
                print("Task updated Successfully!")
        
        elif choice == "4":
            print("------------------------------")
            print("   MARK TASK AS COMPLETED   ")
            print("------------------------------")
            if not tasks:
                print("No tasks available!")
            else:
                for i, task in enumerate(tasks, 1):
                    print(f"{i}. {task['name']}")
                num = int(input("Enter task number to mark as done: ")) - 1
                tasks[num]["done"] = True
                print("Task marked as done Successfully!")
        
        elif choice == "5":
            print("--------------------")
            print("    DELETE TASK     ")
            print("--------------------")
            if not tasks:
                print("No tasks to delete!")
            else:
                for i, task in enumerate(tasks, 1):
                    print(f"{i}. {task['name']}")
                num = int(input("Enter task number to delete: ")) - 1
                removed = tasks.pop(num)
                print(f"Task '{removed['name']}' deleted Successfully!")
        
        elif choice == "6":
            print("Thankyou for using TO-DO list!")
            break
        
        else:
            print("Invalid choice! Try again.")

todo_list()
```

```python
import random  
import string 
def display_menu():
    print("\n")
    print("=======================================")
    print("        PASSWORD GENERATOR APP         ")
    print("=======================================")
    print("  1. Generate New Password             ")
    print("  2. Exit                              ")
    print("=======================================")

def select_complexity():
    print("\n------------------------------------------")
    print("         SELECT PASSWORD COMPLEXITY       ")
    print("------------------------------------------")
    print("  1. Basic   - Letters Only               ")
    print("  2. Medium  - Letters + Numbers          ")
    print("  3. Strong  - Letters + Numbers + Symbols")
    print("------------------------------------------")
    
    complexity = input("  Please Select Complexity (1/2/3) : ") 
    if complexity == "1": 
        characters = string.ascii_letters
        level = "Basic"
    elif complexity == "2":  
        characters = string.ascii_letters + string.digits
        level = "Medium"
    elif complexity == "3": 
        characters = string.ascii_letters + string.digits + string.punctuation
        level = "Strong"
    else:
        print("\n  Invalid selection. Please choose 1, 2, or 3.")
        return None, None 
    
    return characters, level  
def generate_password():
    print("\n------------------------------------------")
    print("           GENERATE NEW PASSWORD          ")
    print("------------------------------------------")
    try:
        length = int(input("  Enter Desired Password Length : ")) 
        
        if length < 4:  
            print("\n  Password length must be at least 4 characters.")
            return
        
        characters, level = select_complexity()  
        
        if characters is None:  
            return
        password = ''.join(random.choice(characters) for _ in range(length))
        print("\n------------------------------------------")
        print("           PASSWORD GENERATED             ")
        print("------------------------------------------")
        print(f"  Password Length     : {length}")
        print(f"  Complexity Level    : {level}")
        print(f"  Generated Password  : {password}")
        print("------------------------------------------")
    
    except ValueError: 
        print("\n  Invalid input. Please enter a valid number for length.")

def password_generator():
    while True: 
        display_menu() 
        choice = input("  Please Select an Option (1-2) : ") 
        
        if choice == "1":  
            generate_password()
        elif choice == "2":   
            print("\n==========================================")
            print("  Thank you for using Password Generator. ")
            print("==========================================\n")
            break 
        else:
            print("\n  Invalid option. Please select 1 or 2.")
password_generator()
```

```python
def calculator():
    print("============================")
    print("     SIMPLE CALCULATOR      ")
    print("============================")
    
    while True:
        print("\n   1. Perform Calculation   ")
        print("   2. Exit                  ")
        
        choice = input("\n Please Select an Option (1-2): ")
        
        if choice == "1":
            print("-----------------------------")
            print("     PERFORM CALCULATION     ")
            print("-----------------------------")
            
            try:
                num1 = float(input("\nEnter first number: "))
                num2 = float(input("Enter second number: "))
                
                print("Operations: + | - | * | /")
                operation = input("\nEnter operation: ")
                
                if operation == "+":
                    result = num1 + num2
                elif operation == "-":
                    result = num1 - num2
                elif operation == "*":
                    result = num1 * num2
                elif operation == "/":
                    if num2 == 0:
                        print("Error: Cannot divide by zero!")
                    else:
                        result = num1 / num2
                        print(f"\nResult: {num1} {operation} {num2} = {result}")
                else:
                    print("Invalid operation!")
                
                if operation in ["+", "-", "*"]:
                    print(f"\nResult: {num1} {operation} {num2} = {result}")
            
            except ValueError:
                print("Invalid input! Please enter numbers only.")
        
        elif choice == "2":
            print("\nThankyou for using Calculator!")
            break
        
        else:
            print("Invalid Choice! Please Enter 1 or 2.")

calculator()
```

```python
import random

user_score = 0
computer_score = 0
tie_score = 0

print("==========================================")
print("       ROCK - PAPER - SCISSORS GAME      ")
print("==========================================")

while True:
    print("\n==========================================")
    print("  1. Play Game                           ")
    print("  2. View Score                          ")
    print("  3. Exit                                ")
    print("==========================================")

    choice = input("  Please Select an Option (1-3) : ")

    if choice == "1":
        print("\n------------------------------------------")
        print("              NEW ROUND                   ")
        print("------------------------------------------")
        print("  Available Choices:")
        print("  1. Rock")
        print("  2. Paper")
        print("  3. Scissors")
        print("------------------------------------------")

        user_input = input("  Enter Your Choice (1/2/3) : ")

        if user_input == "1":
            user_choice = "Rock"
        elif user_input == "2":
            user_choice = "Paper"
        elif user_input == "3":
            user_choice = "Scissors"
        else:
            print("\n  Invalid choice. Please enter 1, 2, or 3.")
            continue

        computer_input = random.randint(1, 3)

        if computer_input == 1:
            computer_choice = "Rock"
        elif computer_input == 2:
            computer_choice = "Paper"
        else:
            computer_choice = "Scissors"

        print("\n------------------------------------------")
        print(f"  Your Choice     : {user_choice}")
        print(f"  Computer Choice : {computer_choice}")
        print("------------------------------------------")

        if user_choice == computer_choice:
            print("  Result          : It is a Tie!")
            tie_score = tie_score + 1
        elif user_choice == "Rock" and computer_choice == "Scissors":
            print("  Result          : You Win!")
            user_score = user_score + 1
        elif user_choice == "Paper" and computer_choice == "Rock":
            print("  Result          : You Win!")
            user_score = user_score + 1
        elif user_choice == "Scissors" and computer_choice == "Paper":
            print("  Result          : You Win!")
            user_score = user_score + 1
        else:
            print("  Result          : Computer Wins!")
            computer_score = computer_score + 1

        print("------------------------------------------")

    elif choice == "2":
        print("\n------------------------------------------")
        print("              SCORE BOARD                 ")
        print("------------------------------------------")
        print(f"  Your Score     : {user_score}")
        print(f"  Computer Score : {computer_score}")
        print(f"  Ties           : {tie_score}")
        print("------------------------------------------")

    elif choice == "3":
        print("\n==========================================")
        print("  Thank you for playing the Game.        ")
        print("==========================================")
        break

    else:
        print("\n  Invalid option. Please select 1, 2, or 3.")
```
